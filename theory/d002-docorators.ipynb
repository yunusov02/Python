## Decorators

Decorators used to modify and extend functionality of another function

```python
def outer_function():

    def inner_function():

        print("Do Something")

    return inner_function


inner_res = outer_function()
res = inner_res()
```

```python
def decorator(func):

    def wrapper(*args, **kwargs):

        print("Do Something before calling function")

        func(*args, **kwargs)

        print("Do Something after calling function")

    return wrapper

@decorator
def greet():
    print("Hello World")


greet()
```
