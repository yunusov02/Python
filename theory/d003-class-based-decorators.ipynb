```python
import time
import functools

class Timer:

    def __init__(self, func):
        self.func = func
        functools.update_wrapper(self, func)

    def __call__(self, *args, **kwargs):

        start = time.perf_counter()

        res = self.func(*args, **kwargs)

        end = time.perf_counter()

        print(f"{self.func.__name__} took, {end - start}s")

        return res


@Timer
def slow_function():
    time.sleep(3)


slow_function()
```
