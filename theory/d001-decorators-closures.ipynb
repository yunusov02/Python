## Passing functions as arguments

````python
def add(a, b):
    return a + b


def multiply(a, b):
    return a * b


def substract(a, b):
    return a - b



def calculation(func, a, b):
    return func(a, b)


def calculation(func, a, b):
    return func(func(a, b), func(a, b))
````

## Functions returning Functions

````python
def create_multiplier(multiplier):

    def multiply(a):
        return a * multiplier

    return multiply


double = create_multiplier(2)
res = double(5)  # This will return 10


print(res)  # Output: 10
````

````python
funcs = []


for i in range(3):

    def func():
        return i

    funcs.append(func)
    print(id(func))  # Print the id of the function object

print([f() for f in funcs])  # Output: [0, 1, 2]
````

````python
funcs = []
for i in range(3):
    def f():
        return i
    funcs.append(f)
    print(id(f))

print(funcs[0] is funcs[1])
print(funcs[1] is funcs[2])
````

````python
funcs = []

for i in range(3):

    def f(param=i):
        return param

    # def f():
    #     return i

    """
    when you use a default parameter (param=i), the current value of i is captured at the time the function is defined.
    This means that each function will have its own copy of the value of i at the time it was created, 
    and they won't be affected by subsequent changes to i.
    """

    funcs.append(f)
    print(id(f))

print("is check:", funcs[0] is funcs[1])
print("values:", [f() for f in funcs])
````
