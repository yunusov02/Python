# Reference Counting vs Garbage Collection

**Reference counting** and **garbage collection (GC)** are both automatic memory management techniques, but they work in fundamentally different ways.

## Reference Counting

Every object keeps a counter of "how many references point to me."

- A new reference is created → counter +1
- A reference is removed → counter -1
- Counter drops to 0 → the object is immediately freed from memory

**Examples:** Python, PHP, and Swift (ARC) use this approach (Python also combines it with a supplementary GC).

**Advantages:**
- Memory is freed immediately (predictable, deterministic)
- No stop-the-world pauses
- Easy to understand

**Disadvantages:**
- **Cyclic references** are a problem: if object A points to B, and B points back to A, neither counter ever reaches 0 — even if nothing else in the program uses them → memory leak
- The counter must be updated on every reference change, which adds overhead (especially in multi-threaded environments, where locking is required)

## Garbage Collection (e.g., mark-and-sweep, generational GC)

The program periodically pauses (or runs a background process) and checks, starting from the "roots," which objects are still reachable. Objects that are unreachable are considered garbage and get cleaned up.

**Examples:** Java (JVM), Go, C#, JavaScript (V8).

**Advantages:**
- No cyclic reference problem — because reachability is checked, not counters
- Often better overall throughput, since there's no per-operation counting cost

**Disadvantages:**
- GC runs can cause stop-the-world pauses — problematic for real-time systems
- Memory isn't freed immediately — the timing is non-deterministic

## Quick Comparison

| | Reference Counting | Garbage Collection |
|---|---|---|
| When memory is freed | Immediately (when counter = 0) | Periodically / later |
| Cyclic references | Problematic (leaks) | Not a problem |
| Overhead | On every reference change | During GC runs (pauses) |
| Predictability | High | Low |

Most modern languages (like Python) use both together: the primary mechanism is reference counting, but a supplementary cycle-detecting GC runs to clean up cyclic references.
