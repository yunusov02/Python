```python
def find_min_max_recursivly(array, n, minimum, maximum):    
    if n < 0:
        return maximum, minimum
    else:
        if array[n] > maximum:
            maximum = array[n]
        elif array[n] < minimum:
            minimum = array[n]

        return find_min_max_recursivly(array, n-1, minimum, maximum)

array = [3, 4, 6, 1, 4, 23]
minimum, maximum = find_min_max_recursivly(array, n=len(array)-1, minimum=array[0], maximum=array[0])
print("Minimum:", minimum)
print("Maximum:", maximum)
```

```python
n = 64

def compute_logarithm_time(n):

    if n == 2:
        return 1

    return 1 + compute_logarithm_time(n // 2)

compute_logarithm_time(n)
```

```python
def uniqeness_element(element, array):

    if not array:
        return True

    if element == array[0]:
        return False

    return uniqeness_element(array[0], array[1:])

array = [1, 4, 234, 2, 6, 34, 54, 5]

uniqeness_element(array[0], array[1:])
```

```python
def multiply(m, n):
    if n == 1:
        return m
    return m + multiply(m, n-1)

multiply(3, 6)
```
